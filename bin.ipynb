Factorial

```python
public class FactorialService {

    @WebMethod(operationName = "operation")
    public String operation(@WebParam(name = "a") final int a) {

        int fact = 1;

        for (int i = 1; i <= a; i++) {
            fact = fact * i;
        }

        return "Factorial of " + a + " is " + fact;
    }
}
```

output jsp for factorial

```python
<%
try {
    client.FactorialService_Service service = new client.FactorialService_Service();
    client.FactorialService port = service.getFactorialServicePort();

    String t1 = request.getParameter("t1");
    int a = Integer.parseInt(t1);

    String result = port.operation(a);
    out.println(result);

} catch (Exception ex) {
    out.println("Error: " + ex.getMessage());
}
%>
```

Prime no prime

```python
public class primenumber {

    @WebMethod(operationName = "prime")
    public String prime(@WebParam(name = "a") final int a) {

        if (a <= 1) {
            return a + "is not a Prime Number";
        }

        for (int i = 2; i <= Math.sqrt(a); i++) {
            if (a % i == 0) {
                return a + "is Not a Prime Number";
            }
        }

        return a + "is a Prime Number";
    }
}
```

output.jsp for prime no prime

```python
<%
try {
    Client.Primenumber_Service service = new Client.Primenumber_Service();
    Client.Primenumber port = service.getPrimenumberPort();

    String t1 = request.getParameter("t1");
    int num = Integer.parseInt(t1);

    String result = port.prime(num);
    out.println("Result: " + result);

} catch (Exception ex) {
    out.println("Error: " + ex.getMessage());
}
%>
```
